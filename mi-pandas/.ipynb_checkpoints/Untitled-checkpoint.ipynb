```python
#Drop nul values
df.dropna()
#if you dont have inplace=true, this actualy returns a new data frame
```

```python
#Remplazar los valores nulos con 0, en vez de borrarlos
#inplace=True is going to actually modify the existing data frame
#not return you a new data frame
df.fillna({"OrderID": 0}, inplace=True)
```

```python
df.rename(columns={"OrderID": "Order ID"}, inplace=True)
```
